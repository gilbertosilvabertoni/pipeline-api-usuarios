```python
import requests
import pandas as pd
```

```python
url = "https://jsonplaceholder.typicode.com/users"

resposta = requests.get(url)

dados = resposta.json()

print(dados)
```

```python
df = pd.DataFrame(dados)

print(df)
```

```python
print(df.columns)
```

```python
print(df["address"][0])
```

```python
df["city"] = df["address"].apply(lambda x: x["city"])
```

```python
print(df[["name", "city"]])
```

```python
df["company_name"] = df["company"].apply(lambda x: x["name"])
```

```python
print(df[["name", "city", "company_name"]])
```

```python
df_limpo = df[["name", "email", "city", "company_name"]]
```

```python
print(df_limpo)
```

```python
df_limpo.to_csv("usuarios.csv", index=False)
```
